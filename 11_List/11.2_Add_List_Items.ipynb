# Add List Items

There are three methods to add items to list: append(), insert() and extend()

### append():

This method appends items to the end of the existing list.

**Example:**

```python
#         [-5]     [-4]    [-3]     [-2]      [-1]     Negative Indexing:
colors = ["Red", "Green", "Blue", "Yellow", "Green"]
#          [0]      [1]     [2]      [3]      [4]      Positive Indexing:

colors.append("green")
print(colors)
```

**What if you want to insert an item in the middle of the list? At a specific index?**

### insert():

This method inserts an item at the given index. User has to specify index and the item to be inserted within the insert() method.

**Example:**

```python
colors = ["voilet", "indigo", "blue"]
#           [0]        [1]      [2]

colors.insert(2, "green")   #inserts item at index 2

# OutPut: colors = ["voilet", "green", "indigo", "blue"]
# indexs              [0]       [1]       [2]      [3]

print(colors)
```

**What if you want to append an entire list or any other collection (set, tuple, dictionary) to the existing list?**

### extend():

This method adds an entire list or any other collection datatype (set, tuple, dictionary) to the existing list.

**Example 1:**

```python
#add a list to a another list
colors = ["voilet", "indigo", "blue"]
rainbow = ["green", "yellow", "orange", "red"]

colors.extend(rainbow)

print(colors)
```

**Example 2:**

```python
#add a tuple to a list
cars = ["Hyundai", "Tata", "Mahindra"]
cars2 = ("Mercedes", "Volkswagen", "BMW") #This is a tuple becouse in this car2 use be ()
cars.extend(cars2)
print(cars)
```

**Example 3:**

```python
#add a set to a list
cars = ["Hyundai", "Tata", "Mahindra"]
cars2 = {"Mercedes", "Volkswagen", "BMW", 23} #This is a Set becouse in this car2 use be {}
cars.extend(cars2)
print(cars)
```

**Example 4:**

```python
#add a dictionary to a list
students = ["Sumit", "Gagan", "Amit"]
students2 = {"Mohit":18, "Aman":19, "Pankaj":19}    #only add keys, does not add values
students.extend(students2)
print(students)
```

### concatenate two lists:

you can simply concatenate two list to join two lists.

**Example:**

```python
colors = ["voilet", "indigo", "blue", "green"]
colors2 = ["yellow", "orange", "red"]
print(colors + colors2)
```
