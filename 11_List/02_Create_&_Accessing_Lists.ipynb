## Creating Lists in Python

To create a list, put a number of expressions in square brackets. That is, use square brackets to indicate the start and end of the list, and separate the items by commas.

### 1. Using Square `[]` Brackets

We can create a list by placing comma `,` separated values within square brackets `[]`.

#### 1.1 We can create empty list object:

```py
a = []
print(a)            # []
print(type(a))      # <class 'list'>
```

#### 1.2 Create list with elements:

```py
a = [1, 2, 3, 'MSK', 23]
print(a)                    # [1, 2, 3, 'MSK', 23]
print(type(a))              # <class 'list'>
```

### 2. Using the `list()` Constructor

You can create a list by passing an iterable (like a string or a range) to the `list()` constructor.

```python
list_from_string = list('Python')  # ['P', 'y', 't', 'h', 'o', 'n']
list_from_range = list(range(5))   # [0, 1, 2, 3, 4]
```

List comprehensions provide a concise way to create lists based on existing lists or ranges.

You can create a list with repeated elements by multiplying a single-element list.

You can start with an empty list and use a loop to append elements.

```python
my_list = []
for i in range(5):
    my_list.append(i)
# Result: [0, 1, 2, 3, 4]
```
