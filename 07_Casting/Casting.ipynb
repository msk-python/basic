># Casting
>
>The conversion of one data type into the other data type is known as type casting in python.
>
>#### Types of Typecasting:
>
>1. [Explicit typecasting](#1-explicit-typecasting-user)
>2. [Implicit typecasting](#2-implicit-typecasting-python)
>___

<br>

>___
>### 1. Explicit typecasting (user):
>The conversion of one data type into another data type, done via developer or programmer's manually as per the requirement, is known as explicit type conversion. 
>
>It can be achieved with the help of Python’s built-in type conversion functions such as **str()**, **int()**, **float()**, **hex()**, **oct()**,etc .
>
>These are some Value for Change Data type:
>```python
>   A = 12      # int
>   B = 2.8     # float
>   C = "3"     # int in string
>   D = "4.2"   # float in string
>```
>
>- [Example 1:](./examples/example1.py) :  **int**, **float**, **int in srting**, **float in string** `to` **String**.
>- [Example 2:](./examples/example2.py) :  **int**, **float**, **int in srting**, **float in string** `to` **Float**.
>- [Example 3:](./examples/example3.py) :  **int**, **float**, **int in srting** `to` **Int** Do not convert (Float in string) `to` **Int** 
>___

<br>

>___
>### 2. Implicit typecasting (python):
>According to the level, one data type is converted into other by the Python interpreter itself (automatically). This is called, implicit typecasting in python.
>
>```py
># Python automatically converts
>A = 12      
>B = 2.8     
>C = 23j
>
>D = A + B + C
>
>print(D , type(D))
>```
>___

````python
A = 12      
B = 2.8     
C = 23j

D = A + B + C
print(D , type(D))
````
