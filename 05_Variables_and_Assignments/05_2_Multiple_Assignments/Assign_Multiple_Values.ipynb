># Assign Multiple Values
>
>### 1. Same value to multiple variables.
>You can assign the `same value` to `multiple variables` in one line:

```python
x = y = z = "Sumit"
print(y)
```

>It will assign value 'Sumit' to all three variables **x**, **y**, **z**. That is, all three labels x, y, z will refer to same location with value 'Sumit'.
>___

>___
>### 2. Multiple Values to Multiple Variables
>Python allows you to assign `multiple values` to `multiple variables` in one line:

```python
x, y, z = 'sumit', 12, 12.5
print(y)
```

>**Note:** Make sure the `number of variables` matches the `number of values`, or else you will get an `error`.
> It will assign the values `orderwise`. like: first variable is geven first value, second variable the second value and so on.
>
>If you want to `swap` values of **x** and **y** you just need to write:

```python
x, y = "Sumit", "Amit"
print(x, y)

x, y = y, x
print(x, y)
```

>___
>### 3. Multiple Values in One Variable
>If you have a `collection of values` in a `list`, `tuple` etc.

```python
fruits = "apple", "banana", "Orange"
print(fruits)

print(id(fruits))
```


>Python allows you to extract the values into variables. This is called **unpacking**.

```python
x , y , z = fruits

# print(id(y))


a=5
b=3
print(id(a))
print(id(b))
```
