### Built-in Modules in Python

Built-in modules are modules that come pre-installed with Python. These modules provide standard functionality that can be used without the need for any additional installations. Here are some of the commonly used built-in modules in Python:

1. **`os`**: Interfaces with the operating system.
2. **`sys`**: Provides access to interpreter variables and functions.
3. **`math`**: Offers mathematical functions and constants.
4. **`datetime`**: Manipulates dates and times.
5. **`time`**: Provides time-related functions.
6. **`random`**: Generates random numbers.
7. **`json`**: Parses and generates JSON.
8. **`re`**: Supports regular expressions.
9. **`urllib`**: Handles URL operations.
10. **`collections`**: Implements specialized container datatypes.
11. **`subprocess`**: Manages subprocesses.
12. **`logging`**: Implements flexible logging.
13. **`functools`**: Higher-order functions for functional programming.
14. **`itertools`**: Provides functions creating iterators for efficient looping.
15. **`heapq`**: Implements a heap queue algorithm.
16. **`threading`**: Provides threading support.
17. **`multiprocessing`**: Supports concurrent processing using processes.
18. **`socket`**: Provides low-level networking interface.
19. **`configparser`**: Handles configuration file parsing.
20. **`csv`**: Supports CSV file operations.
21. **`xml`**: Parses and processes XML.
22. **`hashlib`**: Provides hashing algorithms.
23. **`enum`**: Defines enumerations.
24. **`traceback`**: Prints and formats stack traces.
25. **`pathlib`**: Provides object-oriented filesystem paths.

These modules cover a wide range of functionality and are integral to many Python applications. Each module is well-documented in the [official Python documentation](https://docs.python.org/3/library/index.html), which provides more detailed information and examples for each.
