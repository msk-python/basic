>### 6. Write a program to count the number of even numbers in a list.

```python
total = 0
for number in a :
    if number % 2 == 0:
        total += 1
print("Count of even numbers: ", total)
```

>### 7. Write a program to count the number of odd numbers in a list.

```python
def countOdd(value):
    total = 0
    for number in value :
        if number % 2 != 0:
            total += 1
    print("Count of odd numbers: ", total)

countOdd(a)
```
