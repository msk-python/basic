# Python File Handling

File handling is an important part of any web application.

Python has several functions for creating, reading, updating, and deleting files.

___

## File Open

The key function for working with files in Python is the `open()` function.

The `open()` function takes two parameters; `filename`, and `mode`.

There are four different methods (modes) for opening a file:

`"r"` - Read - Default value. Opens a file for reading, `error` if the file does not exist

`"a"` - Append - Opens a file for appending, `creates` the file if it does not exist

`"w"` - Write - Opens a file for writing, `creates` the file if it does not exist

`"x"` - Create - Creates the specified file, returns an `error` if the file exists

In addition you can specify if the file should be handled as binary or text mode

`"t"` - Text - Default value. Text mode

`"b"` - Binary - Binary mode (e.g. images)

___

## Syntax

To open a file for reading it is enough to specify the name of the file:

```python
open('filename.txt', 'mode') # mode = 'r', 'w', 'a', 'x'
```

## Open a File

To open the file, use the built-in `open()` function.

The `open()` function returns a file `object`, which has a `read()` method for reading the content of the file:

```python
# f = open('sumit') # No such file or directory
f = open('sumit.txt')
print(f)
```

#### Open a file on a different location:

If the file is located in a different location, you will have to specify the file path, like this:

```python
f = open("../sumit.txt", "r")
```

## Read Only

By default the `read()` method returns the whole text, but you can also specify how many characters you want to return:

Return the 16 first characters of the file:

```python
f = open("sumit.txt", "r")  
print( f.read() ) # read the file all content
print( f.read(16) ) # Read only 16 characters
```

## Read Lines

You can return one line by using the `readline()` method:

Read one line of the file:

```python
f = open("sumit.txt")  
f = open("sumit.txt", "r")  
print( f.readline() ) # Read only one line in one time
```

By calling `readline()` two times, you can read the two first lines:

```python
f = open("sumit.txt", "r")  
print( f.readline() ) # Read only one line
print( f.readline() ) # Read only one line
```

You can read the whole file, line by line using By looping:

```python
f = open("sumit.txt", "r")  
for line_no in f:
    print( line_no )
```

## Close Files

It is a good practice to always close the file when you are done with it.

```python
f = open("sumit.txt", "a")  
f.write("\naaaa")
# f.close() # Close the file
```

**Note:** You should always close your files, in some cases, due to buffering, changes made to a file may not show until you close the file.

### Open file  `with`  keyword
not required `close()` function for close file if you are using `with` keyword 

```python
with open("sumit.txt", "r") as f:
    print( f.read() )
```

# Python File Write

## Write to an Existing File

To write to an existing file, you must add a parameter to the `open()` function:

`"a"` - Append - will append to the end of the file

`"w"` - Write - will overwrite any existing content

### Example:

Open the file "sumit.txt" and append content to the file:

```python
f = open("sumit.txt", "a")  
f.write("\naaaa")
f.close()
```

Open the file "sumit.txt" and overwrite the content:

```python
f = open("sumit.txt", "w")  
f.write("I have deleted the content!")
f.close()
```

**Note:** the "w" method will overwrite the entire file.

___

## Create a New File

To create a new file in Python, use the `open()` method, with one of the following parameters:

`"x"` - Create - will create a file, returns an error if the file exist

`"a"` - Append - will create a file if the specified file does not exist

`"w"` - Write - will create a file if the specified file does not exist

### Example

Create a file called "myfile.txt":

```python
f = open("sumit.txt", "a")
f.write(" amit")
f.close()
```

```python
f = open("sumit.txt", "a")
f.write("\namit")
f.close()
```

Result: a new empty file is created!

### Example

Create a new file if it does not exist:

```python
f = open("sumit.py", "w")
f.write("print('Hello Python')")
f.close()
```

## Delete a File

To delete a file, you must import the OS module, and run its `os.remove()` function:

### Example:

Remove the file "sumit.py":

```python
import os
os.remove("sumit.py")
```

## Check if File exist:

To avoid getting an error, you might want to check if the file exists before you try to delete it:

### Example

Check if file exists, _then_ delete it:

```python
import os  

if os.path.exists("sumit.txt"):
    os.remove("sumit.txt")
    print("File Deleted")
else:  
    print("The file does not exist")
```

___

## Delete Folder

To delete an entire folder, use the `os.rmdir()` method:

```python
import os

os.rmdir("SumitFolder") # Remove a folder
os.rmdir("SumitFolder/rohan") # Remove a folder from a folder (/)
os.rmdir("SumitFolder/aman.txt") # Remove a folder in folder
os.remove("SumitFolder/monu.html") # Remove a file from a folder
```

**Note:** You can only remove _empty_ folders.
