# Python Variables
## Variables
Variables are containers for storing data values.

### Creating Variables
Python has no command for declaring a variable.

A variable is created the moment you first assign a value to it.

```python
x = 5
y = "MSK!83"

print(y)

print(x, y)
```

Variables do not need to be declared with any particular type, and can even change type after they have been set.

```python
x = 8+9      # x is type of int
x = "8+9"    # x is now type of str
print(x)
```

## Get the Type
You can get the data type of a variable with the `type()` function.

```python
x = 58
y = "Sumit"
z = 5.8

print(type(x))
print(type(y))
print(type(z))
```

## Single or Double Quotes?
String variables can be declared either by using single or double quotes:

```python
x = 'Amit'
# is the same as
x = "Sumit"

x = """
MSK Institute

Welcome to the MSK Institute of Computer Science and Technology in Shikohabad!

Do you want to learn the skills you need to succeed in today's tech-driven world? If so, then you need to enroll in a computer course at MSK Institute in Shikohabad!
"""

print(x)
```

## Case-Sensitive
Variable names are case-sensitive.

```python
a = 4
A = "MSK!"

# A will not overwrite a

print(a)
print(A)
```
