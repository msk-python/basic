# Built-in Functions in Python

Built-in functions are functions that are always available in Python and can be used without needing to import any modules. They perform common tasks and operations.

## List of Built-in Functions with Examples

Here's a list of some commonly used built-in functions in Python, along with examples:

- math : sum()

```python
sum([.1, .1, .1, .1, .1, .1, .1])       # Output : 0.7
sum([.1, .1, .1, .1, .1, .1, .1, .1])   # Output : 0.7999999999999999

import math
math.fsum([.1, .1, .1, .1, .1, .1, .1, .1]) # Output : 0.8
```

### 1. `abs()`
Returns the `absolute (Positive)` value of a  number.

```python
abs(-5)
# Output: 5
```

### 2. `all()`
Returns `True` if all elements in an iterable are true.

```python
all([True, True, False])
# Output: False
```

### 3. `any()`
Returns `True` if any element in an iterable is true.

```python
any([False, False, True])
# Output: True
```

### 4. `ascii()`
Returns a string containing a printable representation of an object.

```python
ascii("Hello, 世界")
# Output: 'Hello, \u4e16\u754c'
```

### 5. `bin()`
Converts an integer to a binary string.

```python
bin(11)
# Output: '0b1010'
```

### 6. `bool()`
Converts a value to a Boolean (True `1` or False `0`).

```python
bool(0)
# Output: False
```

### 7. `bytes()`
Returns a new "bytes" object.

```python
bytes(6)
# Output: b'\x00\x00\x00\x00\x00'
```

### 8. `callable()`
Checks if an object appears callable.

```python
def sumit():
    print("Hello, Dear")

callable(sumit) # Output: True
callable("sumit") # Output: False
```

### 9. `chr()`
Returns the character that represents the specified Unicode code.

```python
chr(97)
# Output: 'a'
```

### 10. `dict()`
Creates a new dictionary.

```python
dict(name='amit', age=20)
# Output: {'name': 'amit', 'age': 20}
```

### 11. `dir()`
Returns a list of the specified object's properties and methods.

```python
dir(list)
# Output: list of str properties and methods
```

### 12. `divmod()`
Returns a tuple containing the `quotient` and the `remainder` when dividing two numbers.

```python
divmod(28, 2)
# Output: (3, 1)
```

### 13. `enumerate()`
Returns an enumerate object.

```python
e = enumerate(['a', 'b', 'c'])

for index, value in e:
    print(index, value)
# Output:
# 0 a
# 1 b
# 2 c
```

### 14. `eval()`
Evaluates and `executes` an expression.

```python
eval('2 + 2 * 3')
# Output:
```

### 15. `float()`
Returns a `floating-point` number from a number or string.

```python
float('12')
# Output: 12.0
```

### 16. `format()`
Formats a specified value.

```python
format(195, '02x')
# Output: 'ff'
```

### 17. `getattr()`
Returns the value of the specified attribute of an object.

```python
class Person:
    name = 'John'

p = Person()
print(getattr(p, 'name'))
# Output: 'John'
```

### 18. `globals()`
Returns the current global symbol table as a dictionary.

```python
globals()
# Output: dictionary of current global symbols
```

### 19. `hasattr()`
Returns `True` if the specified object has the specified attribute.

```python
class Person:
    name = 'sumit'

p = Person()

hasattr(p, 'name')
# Output: True
```

### 20. `hash()`
Returns the hash value of an object.

```python
hash('hello')
# Output: hash value
```

### 21. `help()`
Invokes the built-in help system.

```python
help(print)
# Displays the help text for the print function
```

### 22. `hex()`
Converts an integer to a hexadecimal string.

```python
hex(255)
# Output: '0xff'
```

### 23. `id()`
Returns the identity of an object.

```python
a = 4
id(a)
# Output: unique identifier for the object
```

### 24. `input()`
Allows user input.

```python
name = input("Enter your name: ")
print(f"Hello, {name}!")
```

### 25. `int()`
Converts a value to an integer.

```python
int('10')
# Output: 10
```

### 26. `isinstance()`
Checks if an object is an instance of a specified class or type.

```python
age = 20

isinstance(age, int)
# Output: True
```

### 27. `issubclass()`
Checks if a class is a subclass of another class.

```python
class A: pass
class B(A): pass

issubclass(B, A)
# Output: True
```

### 28. `iter()`
Returns an iterator object.

```python
my_list = [1, 2, 3]
iterator = iter(my_list)
print(next(iterator))  # Output: 1
print(next(iterator))  # Output: 2
```

### 29. `len()`
Returns the length of an object.

```python
len("hello")
# Output: 5
```

### 30. `list()`
Creates a list.

```python
list('hello')
# Output: ['h', 'e', 'l', 'l', 'o']
```

### 31. `locals()`
Returns the current local symbol table as a dictionary.

```python
def my_func():
    a = 10
    print(locals())

my_func()
# Output: {'a': 10}
```

### 32. `map()`
Applies a function to every item of an iterable.

```python
def square(x):
    return x * x

list(map(square, [1, 2, 3, 4]))
# Output: [1, 4, 9, 16]
```

### 33. `max()`
Returns the largest item in an iterable.

```python
max([1, 2, 3, 4])
# Output: 4
```

### 34. `min()`
Returns the smallest item in an iterable.

```python
min([1, 2, 3, 4])
# Output: 1
```

### 35. `next()`
Retrieves the next item from an iterator.

```python
my_list = [1, 2, 3]
iterator = iter(my_list)


next(iterator)
# Output: 1
```

### 36. `oct()`
Converts an integer to an octal string.

```python
oct(10)
# Output: '0o12'
```

### 37. `open()`
Opens a file and returns a file object.

```python
with open('file.txt', 'r') as file:
    content = file.read()
    print(content)
```

### 38. `ord()`
Returns the Unicode code of a specified character.

```python
ord('a')
# Output: 97
```

### 39. `pow()`
Returns the value of x to the power of y.

```python
pow(2, 3)
# Output: 8
```

### 40. `print()`
Prints a specified message to the screen.

```python
print("Hello, World!")
# Output: Hello, World!
```

### 41. `range()`
Returns a sequence of numbers, starting from 0 and increments by 1 (by default) and ends at a specified number.

```python
for i in range(5):
    print(i)
# Output:
# 0
# 1
# 2
# 3
# 4
```

### 42. `repr()`
Returns a string containing a printable representation of an object.

```python
repr('hello')
# Output: "'hello'"
```

### 43. `reversed()`
Returns a reversed iterator.

```python
print(list(reversed([1, 2, 3])))
# Output: [3, 2, 1]
```

### 44. `round()`
Rounds a number to a specified number of decimals.

```python
round(3.14159, 2)
# Output: 3.14
```

### 45. `set()`
Creates a set.

```python
set([1, 2, 3, 2])
# Output: {1, 2, 3}
```

### 46. `slice()`
Returns a slice object.

```python
my_slice = slice(1, 3)
print([1, 2, 3, 4][my_slice])
# Output: [2, 3]
```

### 47. `sorted()`
Returns a sorted list of the specified iterable.

```python
sorted([3, 1, 2])
# Output: [1, 2, 3]
```

### 48. `str()`
Returns a string object.

```python
str(123)
# Output: '123'
```

### 49. `sum()`
Sums the items of an iterable.

```python
sum([1, 2, 3])
# Output: 6
```

### 50. `tuple()`
Creates a tuple.

```python
tuple([1, 2, 3])
# Output: (1, 2, 3)
```

### 51. `type()`
Returns the type of an object.

```python
type(123)
# Output: <class 'int'>
```

### 52. `vars()`
Returns the `__dict__` attribute of an object.

```python
class Person:
    def __init__(self, name):
        self.name = name

p = Person("Sumit")

print(vars(p))
# Output: {'name': 'Sumit'}
```

### 53. `zip()`
Returns an iterator of tuples, where the first item in each passed iterator is paired together.

```python
names = ['Alice', 'Bob']
ages = [25, 30]
print(list(zip(names, ages)))
# Output: [('Alice', 25), ('Bob', 30)]
```

This list covers many of the built-in functions available in Python. For a complete list, you can refer to the [official Python documentation](https://docs.python.org/3/library/functions.html).
