## Introduction to Lists in Python

In Python, a list is a versatile and powerful data type used to store a collection of items. Lists are one of the most commonly used data structures due to their flexibility and ease of use. They can hold items of various types, including numbers, strings, and even other lists. 

#### Key Characteristics of Lists:

1. **Ordered**: Lists maintain the order of elements as they are added. This means the position of each item is preserved, and you can access items by their index.

2. **Mutable**: You can modify lists after their creation. This includes changing elements, adding new elements, or removing existing ones.

3. **Dynamic**: Lists can grow and shrink in size as needed. You don’t need to specify the size of the list upfront.

4. **Allow Duplicate Values**: Lists can contain multiple occurrences of the same value.

#### Creating a List:

Lists are defined by placing a comma-separated sequence of items within square brackets `[]`. For example:
```python
my_list = [1, 2, 3, 'Python', [4, 5]]
```
In this example, `my_list` contains integers, a string, and another list.

Lists are a fundamental part of Python programming and are used extensively in various applications, from data manipulation to algorithm implementation. Understanding how to work with lists effectively is crucial for any Python developer.

# List Indexes

Each item/element in a list has its own unique index. This index can be used to access any particular item from the list. 

The first item has index `[0]`, second item has index `[1]`, third item has index `[2]` and so on.

## Accessing list items:

### Positive Indexing:

As we have seen that list items have index, as such we can access items using these indexes.

````python
#          [0]      [1]     [2]      [3]      [4]      Positive Indexing:
colors = ["Red", "Green", "Blue", "Yellow", "Green"]

print(colors[3])
````

### Negative Indexing:

Similar to positive indexing, negative indexing is also used to access items, but from the end of the list. 

The last item has index `[-1]`, second last item has index `[-2]`, third last item has index `[-3]` and so on.

````python
colors = ["Red", "Green", "Blue", "Yellow", "Green"]
#         [-5]     [-4]    [-3]     [-2]      [-1]     Negative Indexing:
````

### Check for item:

We can check if a given item is present in the list. This is done using the in keyword.

````python
#         [-5]     [-4]    [-3]     [-2]      [-1]     Negative Indexing:
colors = ["Red", "Green", "Blue", "Yellow", "Green"]
#          [0]      [1]     [2]      [3]      [4]      Positive Indexing:

"Yellow" in colors
````

````python
#         [-5]     [-4]    [-3]     [-2]      [-1]     Negative Indexing:
colors = ["Red", "Green", "Blue", "Yellow", "Green"]
#          [0]      [1]     [2]      [3]      [4]      Positive Indexing:

"sumit" in colors
````

### Range of Index:

You can print a range of list items by specifying where do you want to start, where do you want to end and if you want to skip elements in between the range. 

`Syntax`:

```py
ListName[start : end : jumpIndex]
```

Note: jump Index is optional. We will see this in given examples.

**Example:** printing elements within a particular range:

````python
#           [0]    [1]    [2]     [3]      [4]     [5]      [6]      [7]    [8]
animals = ["cat", "dog", "bat", "mouse", "pig", "horse", "donkey", "goat", "cow"]
#          [-9]   [-8]    [-7]   [-6]     [-5]    [-4]     [-3]     [-2]    [-1]

animals[3:7] 	            # Positive indexes
animals[ -7 : -2 ]	            # Negative indexes
````

Here, we provide index of the element from where we want to start and the index of the element till which we want to print the values.

**Note:** The element of the end index provided will not be included. 

**Example:** printing all element from a given index till the end

````python
#           [0]    [1]    [2]     [3]      [4]     [5]      [6]      [7]    [8]
animals = ["cat", "dog", "bat", "mouse", "pig", "horse", "donkey", "goat", "cow"]
#          [-9]   [-8]    [-7]   [-6]     [-5]    [-4]     [-3]     [-2]    [-1]

print(animals[ 4 : ])	                # Positive indexes
print(animals[ -4 : ])	                # Negative indexes
print(animals[ -2 : -8 ])                # None
````

When no end index is provided, the interpreter prints all the values till the end.

**Example:** printing all elements from start to a given index

````python
#           [0]    [1]    [2]     [3]      [4]     [5]      [6]      [7]    [8]
animals = ["cat", "dog", "bat", "mouse", "pig", "horse", "donkey", "goat", "cow"]
#          [-9]   [-8]    [-7]   [-6]     [-5]    [-4]     [-3]     [-2]    [-1]

print(animals[ : 6 ])	                # Positive indexes
print(animals[ : -5 ])	                # Negative indexes
````

When no start index is provided, the interpreter prints all the values from start up to the end index provided. 

**Example:** print alternate values

````python
#           [0]    [1]    [2]     [3]      [4]     [5]      [6]      [7]    [8]
animals = ["cat", "dog", "bat", "mouse", "pig", "horse", "donkey", "goat", "cow"]
#          [-9]   [-8]    [-7]   [-6]     [-5]    [-4]     [-3]     [-2]    [-1]

print(animals[ 1 : 7 : 2 ])	                # Positive Indexes
print(animals[ -8 : -1 : 2 ])	            # Negative Indexes
print(animals[ -9 : : 2 ])		            # Positive Indexes
````

Here, we have not provided start and index, which means all the values will be considered. But as we have provided a jump index of 2 only alternate values will be printed. 

**Example:** printing every 3rd consecutive withing given range

Here, jump index is 3. Hence it prints every 3rd element within given index.

````python
#           [0]    [1]    [2]     [3]      [4]     [5]      [6]      [7]    [8]
animals = ["cat", "dog", "bat", "mouse", "pig", "horse", "donkey", "goat", "cow"]
#          [-9]   [-8]    [-7]   [-6]     [-5]    [-4]     [-3]     [-2]    [-1]

print(animals[ 1 : 8 : 3 ])    # Positive indexes
````
