>## Python Numbers
>
>As it is clear by the name the Number data types are used to store numeric values.
>
>There are three numeric types in Python:
>
>
>  1. [int](#1-int)
>  1. [float](#2-float)
>  1. [complex](#3-complex)
>___

>___
>## 1. Int
>
>There are two types of integers in Python:
>
>#### 1.1 Integers (signed). 
>
> Integers are represented in Python by `numeric` values `without` decimal point of `unlimited` length. Intergers can be `positive` or `negative`. 
> Integers are whole numbers such as -321, -14, -6, 0, 5, 12, 345 etc.
>
>```py
>a = 12
>b = -17
>```
>
>#### 1.2 Booleans.
>
>These represent the truth values `True` and `False` (not **false** or **true**). 
>
>You can type **bool(1)** or **bool(0)**. Python will return **True** or **False** respectively.
>        
>   * 0: False
>   * 1: True
>```py
>a = bool(0)
>b = bool(1)
>```
>___

````python
a = -23
print(a)

type(a)
````

````python
a = False
print(a)
type(a)
````

>___
>## 2. Float
>
>Float is a number containing **one** or **more** `decimals` **positive** or **negative**.
>
>Float can also be `scientific` numbers with an `"e"` to indicate the power of 10.
>
>For example, 3.14159 is a floating-point number. The `decimal` point signals that it is a floating-point number, not an integer. The number 12 is an integer, but 12.0 is a floating-point number.
>
>* **Decimal Notation:** 3500.75, 0.00005, 147.9101 etc.
>   ```py
>   x = 1.10
>   y = 1.0
>   z = -35.59
>   ```
>
>* **Exponent Notation:** 3.50075E03, 0.5E-04, 1.479101E02 etc.
>   ```py
>   x = 35e3
>   y = 12E4
>   z = -87.7e100
>   ```
>___

````python
a = 35e3
type(a)
````

>___
>## 3. Complex
> 
>A complex number is made up of both real and imaginary components.
>
>If we have a complex number **z**, where **z = a+bi** then **a** would be the real component and **b** would represent hte imaginary component of **z**
>
>real component of **z = 4+3i** is **4** and the imaginary component would be **3**.
>
>Python represents complex numbers in the form **A + Bj**. That is, to represent imaginary number, Python uses **j** (or **J**) in place of traditional **i**.
>
>Complex numbers are written with a "j" as the imaginary part:
>
>```py
>a = 3+5j
>b = 5j
>c = -5j
>```
>
>**Examples:**

````python
x = 1.1 + 3.4j

x = 3 + 4j
print(x)
````

>See, a complex number with non-zero real part is displayed with **parenthese ()** around it.

````python
y = 0 + 4.5j
print(y)
````

>But no parenthese around complex number with real part as zero (0).
