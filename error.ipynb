### AttributeError

Raises AttributeError because 'model' is not defined in the class

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

my_car = Car("Toyota")

my_car.name
```

### EOFError


This might happen in certain IDEs or when input is piped

```python
try:
    data = input("Enter something: ")
except EOFError:
    print("No input provided!")
```

### IOError

Description: Raised when an I/O operation (such as file reading or writing) fails.

```python
try:
    with open('nonexistent_file.txt', 'r') as file:
        content = file.read()
except IOError as e:
    print(e)
```

### IndexError

Description: Raised when a dictionary key is not found in the set of existing keys.

```python
my_dict = {
    'a': 1,
    'b': 2
}
print(my_dict['c']) 
```

### StopIteration
Description: Raised to signal the end of an iterator.

```python
my_iter = iter([1, 2, 3])
while True:
    try:
        print(next(my_iter))
    except StopIteration:
        print("Yes")
        break
```
