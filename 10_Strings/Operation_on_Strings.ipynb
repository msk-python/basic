# Operation on Strings

### Length of a String:

We can find the length of a string using len() function.

**Example:**

```python
fruit = "Mango"
x = len(fruit)
print("Mango is a", x, "letter word.")
```

### String as an Array:

A string is essentially a sequence of characters also called an array. Thus we can access the elements of this array. 

**Example:**

```python
x = "Sumitisgoodboy"

print(x[0:7])
print(x[5])	#returns character at specified index
```

**Note:** This method of specifying the start and end index to specify a part of a string is called slicing. 

**Example:**

```python
x = "Applefruit"
print(x[:5])      #Slicing from Start
print(x[5:])      #Slicing till End
print(x[2:6])     #Slicing in between
print(x[-8:])     #Slicing using negative index
print(x[-5:-1])
```
