># Strings
>
>A string can hold any type of known characters: ***letters***, ***numbers*** and ***special*** characters of any known scripted language.
>
>A Python string is a **sequence** of characters and each character can be individually **accessed** using its **index**.
>
>Strings in python are surrounded by either **single** `('text')` or **double** `("text")` quotation marks.
>
>These are all legal string in Python: 
>
>"abcd", "1234", '$%^&', '??/.<>', "SaEEa", "?#@", "??????", '??????'
>
>You can display a string literal with the `print()` function:
>```py
>print('MSK')
>print("MSK")        # 'MSK' is the same as "MSK".
>```
>___

>## Assign String to a Variable
>Assigning a string to a variable is done with the variable name followed by an equal sign and the string:

````python
a = "MSK"
print(a)
````

Sometimes, the user might need to put quotation marks in between the strings. Example, consider the sentence: He said, “I want to eat an apple”.

How will you print this statement in python?

## Wrong way ❌

````python
print("He said, "I want to eat an apple".")
````

## Right way ✔️

````python
print('He said, "I want to eat an apple".')
print("He said, 'I want to eat an apple'.")
print("He said, '''I want to eat an apple'''.")
print('He said, """I want to eat an apple""".')
print("""He said, 'I want' to "eat" an apple.""")
````

>## Multiline Strings
>
>Multiline strings can be created in two ways:
>
>- **Backslash (\\):**
>    By adding a backslash `(\)` at the `end` before pressing `Enter ↩️`  to continue typing text on the next line in normal single `(')` or double `("")` quote strings. 

````python
about = 'MSK Institute \
    Institute is the best for learn Python \
    in Shikohabad'
````

>- **quotation ( ''', """ )**
>    By typing the text in `triple` quotation marks. (No backslash (\) needed at the end of line). Python allows to type multiline text string by enclosing them in triple quotation marks (both apostrophe `(''')` or quotation `(""")` marks will work).

````python
about = '''MSK Institute
Institute is the best for learn Python
in Shikohabad'''

print(about)
````

````python
a = """
MSK Institute

Welcome to the MSK Institute of Computer Science and Technology in Shikohabad!
Do you want to learn the skills you need to succeed in today's tech-driven world? If so, then you need to enroll in a computer course at MSK Institute in Shikohabad!
"""
print(a)
````
