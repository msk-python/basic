# Python Dictionary Methods

Python has a set of built-in methods that you can use on dictionaries.

| Method | Description |
| --- | --- |
| [clear()](05_Remove_Dictionary_Items.ipynb) | Removes all the elements from the dictionary |
| [copy()]() | Returns a copy of the dictionary |
| [fromkeys()]() | Returns a dictionary with the specified keys and value |
| [get()](02_Access_Dictionary_Items.ipynb) | Returns the value of the specified key |
| [items()](02_Access_Dictionary_Items.ipynb) | Returns a list containing a tuple for each key value pair |
| [keys()](02_Access_Dictionary_Items.ipynb) | Returns a list containing the dictionary's keys |
| [pop()](05_Remove_Dictionary_Items.ipynb) | Removes the element with the specified key |
| [popitem()](05_Remove_Dictionary_Items.ipynb) | Removes the last inserted key-value pair |
| [setdefault()]() | Returns the value of the specified key. If the key does not exist: insert the key, with the specified value |
| [update()](03_Change_Dictionary_Items.ipynb) | Updates the dictionary with the specified key-value pairs |
| [values()](02_Access_Dictionary_Items.ipynb) | Returns a list of all the values in the dictionary |
