# 1. Built-in Data Types

Python offers following buit-in core data types: 

1. Numbers
1. String
1. List
1. Tuple
1. Dictionary
1. etc..

In programming, data type is an **important concept**.


Python has the following data types built-in by default, in these categories:

| Data     |  Type                          |
| -------- | ------------------------------ |
| Text     | str                            |
| Numeric  | int, float, complex            |
| Sequence | list, tuple, range             |
| Mapping  | dict                           |
| Set      | set, frozenset                 |
| Boolean  | bool                           |
| Binary   | bytes, bytearray, memoryview   |
| None     | NoneType                       |

## Setting the Data Type
In Python, the data type is set when you assign a value to a variable:

| Example | Data Type |
| ----------- | ----------- |
| x = "MSK Institute" | str |
| x = 20 | int |
| x = 20.5 | float |
| x = 1j | complex |
| x = ["apple", "banana", "Papaya"] | list |
| x = ("apple", "banana", "Papaya") | tuple |
| x = range(6) | range |
| x = {"name" : "Sumit", "age" : 20} | dict |
| x = {"apple", "banana", "Papaya"} | set |
| x = frozenset({"apple", "banana", "Papaya"}) | frozenset |
| x = True | bool |
| x = b"Hello" | bytes |
| x = bytearray(5) | bytearray |
| x = memoryview(bytes(5)) | memoryview |
| x = None | NoneType |

># 2. `Mutable` and `Immutable`
>
>## 2.1 **Mutable:** 
>The mutable data types are `changeable` their value in place.
>
>Only three types are mutable in python. 
>
>These are: `list`, `dict` and `set`.
>

````python
x = ['apple', 'banana', 23]

print(x)
print(id(x)) 

x[1] = 'sumit'
print(x)

print(id(x)) 
````

>
># 2.2 **Immutable:**
>The immutable data types are `not changeable` their value in place.
>
>These are:  `str`, `int`, `float`, `bool`, `tuple`. 
>___

````python
x = 2

print(x)
print(id(x))

x = 3

print(x)
print(id(x))
````

>___
>## Getting the Data Type
>You can get the data type of any object by using the `type()` function:
>
>Print the data type of the variable x:
>```py
>x = 5
>print(type(x))
>```
